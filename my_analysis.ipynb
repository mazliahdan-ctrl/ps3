```python
# Step 0: Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Load the data
url = "https://web.stanford.edu/class/archive/cs/cs109/cs109.1166/stuff/titanic.csv"
df = pd.read_csv(url)

# Step 2: Save the raw data locally in Colab (optional download)
df.to_csv("titanic.csv", index=False)
# If you want to download to your computer:
from google.colab import files
files.download("titanic.csv")  # triggers download







```

```python

df_clean = df.dropna()
df = df.drop_duplicates()

```

```python
df.head(5)
```

```python
# Convert Age to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

# Remove empty values
df = df.dropna(subset=["Age"])

# Create a new column (for example: decade of age)
df["Age_decade"] = (df["Age"] // 10) * 10



```

```python
age_decade_counts = df["Age_decade"].value_counts().sort_index()
```

```python
plt.figure(figsize=(10,6))
plt.plot(age_decade_counts.index, age_decade_counts.values)

plt.title("Number of Passengers per Age Decade")
plt.xlabel("Age Decade")
plt.ylabel("Number of Passengers")

plt.grid(True)
plt.show()
```
